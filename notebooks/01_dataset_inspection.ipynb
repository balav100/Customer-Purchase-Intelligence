```python
import pandas as pd
import numpy as np
from pathlib import Path

# Resolve dataset path (works whether execution root is workspace root or notebooks folder)
data_path = Path("data/raw/OnlineRetail.csv")
if not data_path.exists():
    data_path = Path("../data/raw/OnlineRetail.csv")

df = pd.read_csv(data_path, encoding="ISO-8859-1")

df_clean = df.copy()

print("Original rows:", len(df))

before = len(df_clean)

df_clean = df_clean.drop_duplicates()

after = len(df_clean)

print("Rows before:", before)
print("Rows after:", after)
print("Duplicates removed:", before - after)

before = len(df_clean)

df_clean = df_clean.dropna(subset=["CustomerID"])

after = len(df_clean)

print("Rows before:", before)
print("Rows after:", after)
print("Rows removed:", before - after)

before = len(df_clean)

df_clean = df_clean[
    ~df_clean["InvoiceNo"].str.startswith("C")
]

after = len(df_clean)

print("Rows before:", before)
print("Rows after:", after)
print("Cancelled transaction rows removed:", before - after)

before = len(df_clean)

df_clean = df_clean[
    df_clean["Quantity"] > 0
]

after = len(df_clean)

print("Rows before:", before)
print("Rows after:", after)
print("Non-positive quantity rows removed:", before - after)

before = len(df_clean)

df_clean = df_clean[
    df_clean["UnitPrice"] > 0
]

after = len(df_clean)

print("Rows before:", before)
print("Rows after:", after)
print("Non-positive price rows removed:", before - after)

#Create Transaction Venue

df_clean["Revenue"] = (
    df_clean["Quantity"] * df_clean["UnitPrice"]
)
df_clean["Revenue"].describe()
print("Non-positive revenue:",
      (df_clean["Revenue"] <= 0).sum())

df_clean["CustomerID"] = df_clean["CustomerID"].astype(int)
df_clean["CustomerID"].dtype

print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicates:", df_clean.duplicated().sum())

print("\nQuantity <= 0:",
      (df_clean["Quantity"] <= 0).sum())

print("UnitPrice <= 0:",
      (df_clean["UnitPrice"] <= 0).sum())

print("Revenue <= 0:",
      (df_clean["Revenue"] <= 0).sum())

print("\nUnique customers:",
      df_clean["CustomerID"].nunique())

print("Unique invoices:",
      df_clean["InvoiceNo"].nunique())

print(df_clean.isnull().sum())
print("Total missing values:", df_clean.isnull().sum().sum())
print(df_clean.dtypes)
print(df_clean.head())

df_clean.to_csv(
    "../data/processed/online_retail_clean.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
